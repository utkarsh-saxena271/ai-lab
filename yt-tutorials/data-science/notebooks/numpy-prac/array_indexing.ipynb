```python
import numpy as np
```

Array Indexing and Slicing

```python
arr = np.arange(11,21)
arr
```

```python
arr[6]
```

```python
arr[1:3]
```

Numpy Indexing and Slicing of Matrix

```python
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
arr[0]
```

```python
arr[0][1]
```

```python
arr[0,0]
```

```python
arr[0:6, 0:1]
```

Boolean Indexing

```python
arr = np.arange(11,21)
arr
```

```python
bool_index = arr%2 == 0
bool_index
```

```python
arr = arr[bool_index]
arr
```
